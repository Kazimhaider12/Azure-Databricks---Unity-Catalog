# Delta Lake

```python
%sql
create table ext_cata.ext_schema.deltatable(
    id int,
    name string,
    city string
)using delta
location 'abfss://mycontainer@kazimdatabrickstorage.dfs.core.windows.net/deltalake/deltatable'
```

```python
%sql
ALTER TABLE ext_cata.ext_schema.deltatable
SET TBLPROPERTIES ('delta.enableDeletionVectors' = 'false');
```

```python
%sql
insert into ext_cata.ext_schema.deltatable values
(1, 'kazim', 'lahore'),
(2, 'ilal', 'lahore'),
(3, 'Ahmad', 'lahore')
```

```python
%sql
describe extended ext_cata.ext_schema.deltatable
```

```python
%sql
select * from ext_cata.ext_schema.deltatable
```

```python
%sql
update ext_cata.ext_schema.deltatable
set name = 'haider'
where id = 1 
```

```python
%sql
select * from ext_cata.ext_schema.deltatable
```

```python
%sql
describe history ext_cata.ext_schema.deltatable
```

```python
%sql
restore table ext_cata.ext_schema.deltatable to version as of 3
```

```python
%sql
select * from ext_cata.ext_schema.deltatable
```
